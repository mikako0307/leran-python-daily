```python
from bs4 import BeautifulSoup
from urllib import request

url = 'https://www.aozora.gr.jp/cards/000148/files/2371_13943.html'
response = request.urlopen(url)
soup = BeautifulSoup(response, "html.parser")
response.close()

print(soup)
```

```python
main_text = soup.find('div', class_='main_text')
print(main_text)
```

```python
tags_to_delete = main_text.find_all(['rp', 'rt'])
for tag in tags_to_delete:
  tag.decompose()
print(main_text)
```

```python
main_text = main_text.get_text()
print(main_text)
```

```python
import re
main_text = re.sub(r"[\u3000\r\n]", "", main_text)
print(main_text)
```

```python
import re
import unicodedata

# 句読点は消さない！改行と全角スペースだけ除去
main_text_clean = re.sub(r"[\u3000\r\n]", "", main_text)

def normalize_text(text):
    text = unicodedata.normalize("NFKC", text)
    # 青空文庫の旧字体を最低限現代化（必要に応じて追加）
    text = text.replace("ゐ", "い").replace("ゑ", "え")
    text = text.replace("既", "既").replace("突", "突")
    text = text.replace("難", "難").replace("海", "海")
    text = text.replace("――", "—")
    return text

main_text_clean = normalize_text(main_text_clean)

print(main_text_clean[:300])

```

```python
from transformers import pipeline

nlp = pipeline(
    "text-generation",
    model="llm-jp/llm-jp-3-1.8b-instruct",
    tokenizer="llm-jp/llm-jp-3-1.8b-instruct"
)

prompt = f"以下の文章を200文字以内で要約してください。\n\n{main_text_clean}\n\n要約："

result = nlp(prompt, max_length=300, do_sample=False)
print(result[0]["generated_text"])
```
