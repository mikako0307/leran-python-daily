```python
#学習データ(train.csv)とテストデータ(test.csv)のデータフレームへの読み込み
train = pd.read_csv('/kaggle/input/titanic/train.csv')
test = pd.read_csv('/kaggle/input/titanic/test.csv')
```

```python
#Title→ rare → one-hot
rare_titles = ['Dr','Rev','Mlle','Major','Col','the Countess','Capt','Ms','Sir','Lady','Mme','Don','Jonkheer']

for df in [train, test]:
    df['Title_tr'] = df['Name'].map(lambda x: x.split(', ')[1].split('. ')[0])
    df['Title_tr'] = df['Title_tr'].replace(rare_titles, "Others")
    df['Title_tr'] = df['Title_tr'].apply(lambda x: x if x in ['Master','Miss','Mr','Mrs','Others'] else 'Others')

title_dummies_train = pd.get_dummies(train['Title_tr'], prefix='Title')
title_dummies_test  = pd.get_dummies(test['Title_tr'],  prefix='Title')

title_dummies_test = title_dummies_test.reindex(columns=title_dummies_train.columns, fill_value=0)

train = pd.concat([train, title_dummies_train], axis=1)
test  = pd.concat([test,  title_dummies_test],  axis=1)
```

```python
# ① train の各レベルの中央値を計算
age_medians_3 = train.groupby(['Pclass', 'Sex', 'Title_tr'])['Age'].median()
age_medians_2 = train.groupby(['Pclass', 'Sex'])['Age'].median()
age_medians_1 = train.groupby(['Pclass'])['Age'].median()
age_median_all = train['Age'].median()

# ② 補完関数
def fill_age(row):
    p, s, t = row['Pclass'], row['Sex'], row['Title_tr']
    
    # 3変数
    if (p, s, t) in age_medians_3.index:
        return age_medians_3.loc[(p, s, t)]
    
    # 2変数
    if (p, s) in age_medians_2.index:
        return age_medians_2.loc[(p, s)]
    
    # 1変数
    if p in age_medians_1.index:
        return age_medians_1.loc[p]
    
    # fallback
    return age_median_all

# ③ train の Age 補完
train['Age'] = train.apply(
    lambda row: fill_age(row) if pd.isnull(row['Age']) else row['Age'],
    axis=1
)

# ④ test の Age 補完
test['Age'] = test.apply(
    lambda row: fill_age(row) if pd.isnull(row['Age']) else row['Age'],
    axis=1
)

# ⑤ Age_bin（ビン化）
age_bins = [0, 12, 20, 40, 60, 100]

for df in [train, test]:
    df['Age_bin'] = pd.cut(df['Age'], bins=age_bins, labels=False, include_lowest=True)
```

```python
# Fare の補完
test['Fare'] = test['Fare'].fillna(test['Fare'].median())

# Fare_bin（train の境界で test も切る）
fare_bins = pd.qcut(train['Fare'], 4, retbins=True)[1]

train['Fare_bin'] = pd.cut(train['Fare'], bins=fare_bins, labels=False, include_lowest=True)
test['Fare_bin']  = pd.cut(test['Fare'],  bins=fare_bins, labels=False, include_lowest=True)
```

```python
# Ticket の頭文字を抽出
for df in [train, test]:
    df['Ticket_Letter'] = df['Ticket'].astype(str).str[0]

# カテゴリ → 数値
ticket_map = {v: i for i, v in enumerate(train['Ticket_Letter'].unique())}
train['Ticket_Letter'] = train['Ticket_Letter'].map(ticket_map)
test['Ticket_Letter']  = test['Ticket_Letter'].map(ticket_map)
```

```python
# Cabin の頭文字を抽出（欠損は 'U' とする）
for df in [train, test]:
    df['Cabin_Deck'] = df['Cabin'].astype(str).str[0]
    df['Cabin_Deck'] = df['Cabin_Deck'].replace('n', 'U')  # NaN → 'U'

# カテゴリ → 数値
deck_map = {v: i for i, v in enumerate(train['Cabin_Deck'].unique())}
train['Cabin_Deck'] = train['Cabin_Deck'].map(deck_map)
test['Cabin_Deck']  = test['Cabin_Deck'].map(deck_map)
```

```python
#Family size
for df in [train, test]:
    df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
    df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

for df in [train, test]:
    df['SibSp_bin'] = pd.cut(df['SibSp'], bins=[-1, 0, 1, 2, 8], labels=False)
    df['Parch_bin'] = pd.cut(df['Parch'], bins=[-1, 0, 1, 2, 6], labels=False)
    df['FamilySize_bin'] = pd.cut(df['FamilySize'], bins=[0, 1, 2, 4, 11], labels=False)
```

```python
# Embarked の補完
train['Embarked'] = train['Embarked'].fillna(train['Embarked'].mode()[0])

```

```python
embarked_map = {'S': 0, 'C': 1, 'Q': 2}

train['Embarked'] = train['Embarked'].map(embarked_map)
test['Embarked'] = test['Embarked'].map(embarked_map)
```

```python
for df in [train, test]:
    df['Ticket_is_numeric'] = df['Ticket'].str.isdigit().astype(int)
    df['HasCabin'] = (df['Cabin_Deck'] != deck_map['U']).astype(int)
```

```python
from lightgbm import LGBMClassifier
from sklearn.model_selection import StratifiedKFold


import numpy as np
```

```python
features = [
    'Pclass', 'Sex', 'Embarked',
    'FamilySize', 'IsAlone',
    'Age_bin', 'Fare_bin',
    'SibSp_bin', 'Parch_bin', 'FamilySize_bin',
    'Ticket_Letter', 'Ticket_is_numeric',
    'Cabin_Deck', 'HasCabin'
] + list(title_dummies_train.columns)

X = train[features]
y = train['Survived']

kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

preds = np.zeros(len(test))

for train_idx, val_idx in kf.split(X, y):
    X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]

    model = LGBMClassifier(
        objective='binary',
        n_estimators=800,
        learning_rate=0.03,
        num_leaves=31,
        max_depth=1,
        min_child_samples=10,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    )

    model.fit(X_train, y_train)
    preds += model.predict_proba(test[features])[:, 1] / 5

# 0/1 に変換
pred = (preds >= 0.5).astype(int)

```

```python
submission = pd.DataFrame({
    'PassengerId': test['PassengerId'],
    'Survived': pred
})

submission.to_csv('submission.csv', index=False)
```
